```python
import re

import numpy as np
import pandas as pd


def _to_snake_case(name: str) -> str:
    """Convert a raw CSV header like 'GRADUATION RATE STATUS' or
    'Attendance Rate (%)' into 'graduation_rate_status' / 'attendance_rate'.
    Collapses any run of non-alphanumeric characters (spaces, %, &, /, etc.)
    into a single underscore.
    """
    name = re.sub(r"[^0-9a-zA-Z]+", "_", name.strip())
    return name.strip("_").lower()
```

```python
PLACEMENT_ORDER = [
    "Emerging K", "Early K", "Mid K", "Late K",
    "Emerging Grade 1", "Early Grade 1", "Mid Grade 1", "Late Grade 1",
    "Grade 2", "Grade 3", "Grade 4", "Grade 5",
    "Grade 6", "Grade 7", "Grade 8",
]


def clean_diagnostics(df: pd.DataFrame) -> pd.DataFrame:
    """Clean the student diagnostic results extract.

    Steps:
      1. Split 'Overall Placement & Scale Score' (e.g. "Grade 7 (588)") into
         a placement label and a numeric score, and validate the extracted
         score against the already-separate 'Scale Score' column instead of
         blindly trusting either one.
      2. Build an ordered 'placement_grade_rank' category so placements can
         be sorted/compared instead of treated as arbitrary strings.
      3. Parse 'Diagnostic Date' into a real datetime.
      4. Coerce numeric columns with pd.to_numeric(errors="coerce") rather
         than a plain astype(), so any unexpected non-numeric entry becomes
         NaN (visible/checkable later) instead of crashing the whole script.
      5. Compute two derived metrics: growth_gap and behavior_index.
    """
    out = df.copy()
    out.columns = [_to_snake_case(c) for c in out.columns]

    # --- 1. Split + validate the combined placement/score column ---
    # Regex pulls "Grade 7 (588)" apart into ("Grade 7", "588").
    extracted = out["overall_placement_scale_score"].str.extract(
        r"^(?P<placement>.+?)\s*\((?P<paren_score>\d+)\)$"
    )
    out["placement"] = extracted["placement"]
    paren_score = pd.to_numeric(extracted["paren_score"], errors="coerce")

    # Validate rather than assume: the parenthesized number should always
    # match the separate 'scale_score' column. We don't raise on mismatch
    # (a single bad row shouldn't kill the whole pipeline) - we just report
    # how many rows disagree so a human can investigate.
    mismatches = (paren_score != out["scale_score"]).sum()
    if mismatches:
        print(f"[clean_diagnostics] WARNING: {mismatches} rows where the "
              f"placement score doesn't match Scale Score - investigate.")

    # The parenthesized score is now redundant with scale_score, so drop the
    # original compound column entirely rather than keep two representations
    # of the same number around.
    out = out.drop(columns=["overall_placement_scale_score"])

    # --- 2. Ordered category for placement ---
    out["placement_grade_rank"] = pd.Categorical(
        out["placement"], categories=PLACEMENT_ORDER, ordered=True
    )
    unmapped = out.loc[out["placement_grade_rank"].isna() & out["placement"].notna(), "placement"]
    if not unmapped.empty:
        print(f"[clean_diagnostics] WARNING: {unmapped.nunique()} placement "
              f"label(s) not found in PLACEMENT_ORDER: {sorted(unmapped.unique())}")

    # --- 3. Parse date ---
    out["diagnostic_date"] = pd.to_datetime(
        out["diagnostic_date"], format="%m/%d/%y", errors="coerce"
    )

    # --- 4. Coerce numeric columns ---
    numeric_cols = [
        "scale_score", "typical_growth", "stretch_growth",
        "attendance_rate", "tardies_ytd", "behavior_incidents_ytd",
        "positive_behavior_points", "engagement_rating_1_5",
    ]
    for col in numeric_cols:
        out[col] = pd.to_numeric(out[col], errors="coerce")

    # --- 5. Derived metrics ---
    # growth_gap: how much more (or less) growth is needed to hit the
    # "Stretch" target versus the "Typical" target. Positive = stretch goal
    # asks for more growth than typical; useful for flagging ambitious goals.
    out["growth_gap"] = out["stretch_growth"] - out["typical_growth"]

    # behavior_index: a simple composite score - positive behavior points
    # minus a penalty for incidents (weighted heavier) and tardies (weighted
    # lighter). This is intentionally simple/transparent rather than a
    # "black box" formula, so a teammate can tune the weights later.
    out["behavior_index"] = (
        out["positive_behavior_points"]
        - (out["behavior_incidents_ytd"] * 5)
        - (out["tardies_ytd"] * 1)
    )

    return out
```

```python
# The measure "families" in this file. Each has its own STATUS column that
# is blank whenever the measure doesn't apply to that row (wrong school
# level) OR was suppressed by the state for small subgroup size. We add one
# is_suppressed flag per family so cleaning code downstream never has to
# reinterpret raw NaNs, and never invents a rate for a suppressed cell.
RATING_FAMILIES = {
    "reading_math": "reading_math_indicator_status",
    "science_social_studies_combined_writing": "science_social_studies_combined_writing_status",
    "graduation_rate": "graduation_rate_status",
    "english_learners_progress": "english_learners_progress_status",
    "quality_of_school_climate_and_safety": "quality_of_school_climate_and_safety_status",
    "postsecondary": "postsecondary_status",
}


def clean_school_summary(df: pd.DataFrame) -> pd.DataFrame:
    """Clean the KYRC25 ACCT school accountability summary extract.

    Steps:
      1. Rename all columns to snake_case for consistent, code-friendly names.
      2. Cast every rate/status/rating column to float via
         pd.to_numeric(errors="coerce"). Blanks become NaN (not 0) because a
         blank here means "no rate was published for this cell" - filling
         with 0 would make a suppressed/inapplicable cell look like the
         school scored zero, which is factually wrong and would wreck any
         average.
      3. Add one is_suppressed_<family> boolean per measure family, so
         downstream analysis can filter/report on suppression explicitly
         instead of silently dropping or averaging over NaNs.
    """
    out = df.copy()
    out.columns = [_to_snake_case(c) for c in out.columns]

    # --- 2. Coerce all the measure columns (everything after 'level') ---
    level_idx = out.columns.get_loc("level")
    rate_cols = out.columns[level_idx + 1:]
    out[rate_cols] = out[rate_cols].apply(pd.to_numeric, errors="coerce")

    # --- 3. Suppression flags, one per rating family ---
    # NOTE: this flags "no status published" which conflates true small-cell
    # suppression with "measure doesn't apply to this school level" (e.g.
    # graduation rate on an elementary school). See profiling_notes.md -
    # refining this to only flag *applicable* rows is a follow-up.
    for family_name, status_col in RATING_FAMILIES.items():
        out[f"is_suppressed_{family_name}"] = out[status_col].isna()

    return out
```

```python
def clean_school_improvement(df: pd.DataFrame) -> pd.DataFrame:
    """Clean the KYRC25 ACCT school improvement (federal classification) extract.

    Steps:
      1. Rename columns to snake_case.
      2. Cast 'number_of_years_identified' to a nullable Int64 (not a plain
         int and not filled with 0) because a blank here means the school is
         newly identified this year and doesn't have a year count yet - that
         is a different fact than "zero years," so we preserve it as missing
         rather than guessing a value.
      3. Standardize the federal classification text (trim whitespace,
         uppercase) so downstream filtering/grouping ("ATSI" vs " atsi ")
         doesn't silently miss rows due to formatting differences.
    """
    out = df.copy()
    out.columns = [_to_snake_case(c) for c in out.columns]

    # --- 2. Nullable integer, NaN preserved on purpose (see docstring) ---
    out["number_of_years_identified"] = pd.to_numeric(
        out["number_of_years_identified"], errors="coerce"
    ).astype("Int64")

    # --- 3. Standardize classification text ---
    out["federal_classification"] = (
        out["federal_classification"].str.strip().str.upper()
    )

    return out
```
