```python
csv_path = "combined_files.csv"
if not os.path.exists(csv_path):
    possible_paths = ["data_cleaned/combined_files.csv"] + glob.glob("**/combined_files.csv", recursive=True)
    for path in possible_paths:
        if os.path.exists(path):
            csv_path = path
            break

if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
else:
    print(f"Warning: '{csv_path}' not found. df set to empty DataFrame.")
    df = pd.DataFrame()
```

```python
if 'Overall Placement & Scale Score' in df.columns:
    placement = df['Overall Placement & Scale Score'].astype(str).str.replace(r' \(\d+\)$', '', regex=True)
else:
    placement = pd.Series(pd.NA, index=df.index)
    print("Warning: 'Overall Placement & Scale Score' not found in df; 'grade_band' will be empty.")

def band(p):
    if pd.isna(p):
        return pd.NA
    if 'K' in p:
        return 'K'
    grade = int(p.split()[-1])
    if grade <= 2:
        return '1-2'
    elif grade <= 5:
        return '3-5'
    else:
        return '6-8'
```

```python
if df.empty:
    print("df is empty. Check that 'combined_files.csv' was loaded successfully.")
else:
    df.columns = df.columns.str.strip()
    required = ['Scale Score', 'Attendance Rate (%)', 'Absenteeism Rate (%)']
    missing = [c for c in required if c not in df.columns]
    if missing:
        print("Missing columns:", missing)
        print("Available columns:", df.columns.tolist())
    else:
        print(df[required].describe())
```

```python
if df.empty:
    print("df is empty. Cannot compute correlation heatmap.")
else:
    df.columns = df.columns.str.strip()
    required = ['Absenteeism Rate (%)', 'Scale Score']
    missing = [c for c in required if c not in df.columns]
    if missing:
        print("Missing columns:", missing)
        print("Available columns:", df.columns.tolist())
    else:
        corr = df[required].corr()
        sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
        plt.show()
```

```python
if df.empty:
    print("df is empty. Cannot plot Scale Score histogram.")
else:
    df.columns = df.columns.str.strip()
    if 'Scale Score' not in df.columns:
        print("Missing column 'Scale Score'. Cannot plot histogram.")
    else:
        sns.histplot(df['Scale Score'], kde=True)
        plt.title(f"Skew = {df['Scale Score'].skew():.2f}")
        plt.show()
```
